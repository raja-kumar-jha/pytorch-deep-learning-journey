### Random Tensors

Random number are important because neural network start learning with full of random numbers and then it adjust those numbers to better represent data.



`start with random numbers -> look at data -> update random number -> again look at the data -> again update the data and keep doing until get the result`



```python
# Create the rendom tensor
random_tensor = torch.rand(2,3)
random_tensor
```

```python
random_tensor.ndim
```

```python
random_tensor.shape
```

```python
# Create a random tensor with similar shape of an image tensor
random_image_size_tensor = torch.rand(size=(224, 224, 3)) # heigh, width, colour channels (R, G, B)
random_image_size_tensor
```

```python
random_image_size_tensor.ndim
```

```python
random_image_size_tensor.shape
```

### Creating a range of tensors and tensor-like

```python
# use torch.range()
torch.arange(0, 10)
```

```python
one_to_ten = torch.arange(start=1, end=11, step=2)
one_to_ten
```

```python
# Creating tensors like
ten_zeros = torch.zeros_like(input=one_to_ten)
ten_zeros
```

### Tensor datatype

```python
float_32_tensor = torch.tensor([3.0, 4.5, 5.6],
                               dtype=None,
                               )
float_32_tensor
```

```python
float_32_tensor.dtype
```

```python
float_64_tensor = torch.tensor([2.3, 4.9, 9.7],
                               dtype=torch.float64, # what datatype is the tensor(e.g. float64, flaot32, float16)
                               device=None, # what device is your tensor on
                               requires_grad=False) # whether or not to track gradients with this tensors operations
float_64_tensor
```

```python
# conversion
flaot_16_tensor = float_32_tensor.type(torch.float16)
flaot_16_tensor
```

### Getting information from tensors

### Manipulating Tensors (tensors Opertaions)

Tensors Operations:
* Addition
* Substraction
* Multiplication(Element-wise)
* Division
* Matrix Multipication

```python
# Create a tensor and add 101
tensor = torch.tensor([1, 2, 3])
tensor + 101
```

```python
torch.add(tensor, 101)
```

```python
# Substarct 10 from tensor
tensor - 12
```

```python
# Multiply tensor by 10\
tensor * 10
```

```python
torch.mul(tensor, 10)
```

```python
# divide tensor by 10
tensor / 10
```

```python
torch.div(tensor, 10)
```
